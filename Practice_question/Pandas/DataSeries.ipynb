1.  Creating a Series

```python
import pandas as pd
a=(1,2,3,4,5)
print("Tuple :",a)
print(type(a))# type(a) use for builtin data type in python
st=pd.Series(a)
print(st.dtype)# st.dtype use for Pandas Series 
print("Series :",st)
```

2. Series to List Conversion

```python
print(st)
#to list
l=st.tolist() # convert series to list
print(type(l))
print("list :",l)
```

3. Series Arithmetic

```python
s1=pd.Series((2,4,6,8,10))
s2=pd.Series((1,3,5,7,9))
#Arithmetic method use like + , -, *  ,/ 
print("Add :\n",s1+s2)
print("sub :\n",s1-s2)
print("mul\n",s1*s2)
print("div\n",s1/s2)

```

4. Series Comparison

```python
print(s1>s2)
print(s1<s2)
print(s1==s2)
```

5. Dict to Series

18. Five-Number Summary

```python
s = pd.Series(np.random.uniform(0, 20, 20))
print("Five number summary :")
print("min :",s.min())
print("25th :",s.quantile(.25))
print("median :",s.median())
print("75th :",s.quantile(.75))
print("max :",s.max())
```

27. Date Strings to TimeSeries

```python
import pandas as pd
import datetime
s = pd.Series([
    "01 Jan 2015",
    "10-02-2016",
    "20180307",
    "2014/05/06",
    "2016-04-12",
    "2019-04-06T11:20"
])

print("Original Series:")
print(s)
print(pd.DatetimeIndex(s))
```

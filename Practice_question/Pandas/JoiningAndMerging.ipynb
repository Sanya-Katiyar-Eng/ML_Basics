#### Joining and merging DataFrame

1. Join DataFrames along Rows (Join Rows)

```python
import pandas as pd

# Student DataFrame 1
student_data1 = pd.DataFrame({
    "student_id": ["S1", "S2", "S3", "S4", "S5"],
    "name": [
        "Danniella Fenton",
        "Ryder Storey",
        "Bryce Jensen",
        "Ed Bernal",
        "Kwame Morin"
    ],
    "marks": [200, 210, 190, 222, 199]
})

# Student DataFrame 2
student_data2 = pd.DataFrame({
    "student_id": ["S4", "S5", "S6", "S7", "S8"],
    "name": [
        "Scarlette Fisher",
        "Carla Williamson",
        "Dante Morse",
        "Kaiser William",
        "Madeeha Preston"
    ],
    "marks": [201, 200, 198, 219, 201]
})

# Display DataFrames
print("student_data1:")
print(student_data1)

print("\nstudent_data2:")
print(student_data2)
```

```python
pd.concat([student_data1,student_data2])
```

2. Join DataFrames along Columns (Join Columns)

```python
pd.concat([student_data1,student_data2],axis=1)
```

5. Join Rows and Merge with Another DataFrame

```python
df=pd.concat([student_data1,student_data2])
print(df)
# creatin df
df3 = pd.DataFrame({
    "student_id": ["S9"],
    "name": ["Sanya"],
    "marks": [210]
})

df1=pd.concat([df,df3])
print(df1)
```

7. Join with Matching Records (Inner Join)

```python
df=pd.merge(student_data1,student_data2,on="student_id",how="inner")
df
```

8. Left Join Using Keys from Left DataFrame

```python
#using left
pd.merge(student_data1,student_data2,on="student_id",how="left")
```

```python
#using right
pd.merge(student_data2,student_data1,on="student_id",how="left")
```

9. Right Join Using Keys from Right DataFrame

```python
#using right
pd.merge(student_data2,student_data1,on="student_id",how="right")
```

```python
#using left
pd.merge(student_data1,student_data2,on="student_id",how="right")
```

10. Merge Using Multiple Join Keys

```python
pd.merge(student_data1,student_data2,on=["student_id","name"])
```

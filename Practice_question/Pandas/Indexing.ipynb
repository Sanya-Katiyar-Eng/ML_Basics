### Pandas Indexing

1. Display Default and Set Column as Index

```python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "school_code": ["s001", "s002", "s003", "s001", "s002", "s004"],
    "class": ["V", "V", "VI", "VI", "V", "VI"],
    "name": [
        "Alberto Franco",
        "Gino Mcneill",
        "Ryan Parkes",
        "Eesha Hinton",
        "Gino Mcneill",
        "David Parkes"
    ],
    "date_of_birth": [
        "15/05/2002",
        "17/05/2002",
        "16/02/1999",
        "25/09/1998",
        "11/05/2002",
        "15/09/1997"
    ],
    "age": [35, 32, 33, 30, 31, 32],
    "address": [
        "street1",
        "street2",
        "street3",
        "street1",
        "street2",
        "street4"
    ],
    "teacher_id": ["t1", "t2", "t3", "t4", "t5", "t6"]
})

print(df)

print("Set school_code as index")
print(df.set_index(df["school_code"]))

print("set name as index : ")
df.set_index(df["name"])
```

5. DataFrame with Interval Index

```python
import pandas as pd
print("Create an Interval Index using IntervalIndex.from_breaks:")
df_interval = pd.DataFrame({"X":[1, 2, 3, 4, 5, 6, 7]},
                            index = pd.IntervalIndex.from_breaks(
                            [0, 0.5, 1.0, 1.5, 2.0, 2.5, 3, 3.5]))    
print(df_interval)
print(df_interval.index)

print("\nCreate an Interval Index using IntervalIndex.from_tuples:")
df_interval = pd.DataFrame({"X":[1, 2, 3, 4, 5, 6, 7]},             
                            index = pd.IntervalIndex.from_tuples(
                            [(0, .5), (.5, 1), (1, 1.5), (1.5, 2), (2, 2.5), (2.5, 3), (3, 3.5)]))
print(df_interval)
print(df_interval.index)

print("\nCreate an Interval Index using IntervalIndex.from_arrays:")
df_interval = pd.DataFrame({"X":[1, 2, 3, 4, 5, 6, 7]},             
                            index = pd.IntervalIndex.from_arrays(
                            [0, .5, 1, 1.5, 2, 2.5, 3], [.5, 1, 1.5, 2, 2.5, 3, 3.5]))
print(df_interval)
print(df_interval.index) 
```

#### Using Library sklearn

```python
#  Simple Linear Regression Example
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
cgpa = np.round(np.random.uniform(5.0, 10.0, 200), 2)
noise = np.random.normal(0, 0.6, 200)
package = 0.8 * cgpa - 2 + noise
package = np.clip(package, 1.5, None)
package = np.round(package, 2)
df=pd.DataFrame({"cgpa":cgpa,
                 "Package":package})
df
```

```python
sns.scatterplot(x="cgpa",y="Package",data=df)
```

```python
x=df.iloc[:,0:1]
y=df.iloc[:,-1]
x,y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train )
```

```python
x_test
```

```python
y_test
```

```python
lr.predict(x_test.iloc[0].values.reshape(1,1))
```

```python
lr.predict(x_test.iloc[1].values.reshape(1,1))
```

```python

sns.scatterplot(x="cgpa",y="Package",data=df)
plt.plot(x_test,lr.predict(x_test),color="red")
```

```python
# Slove value..............................
m=lr.coef_
m
```

```python
b=lr.intercept_
# offset
```

```python
# y=mx+b    aese aati hai package ka value nikalne ke liye

print((m*7.90)+b)
```

### Without using library

```python
class LR:
    def __init__(self):
        self.m=None
        self.c=None
    def fit(self,x_train,y_train):
        num=0
        den=0
        for i in range(x_train.shape):
            num=num+(x_train[i]-np.mean(x_train))*(y_train[i]-np.mean(y_train))
            den=den+(x_train[i]-np.mean(x_train))**2
        self.m=num/den
        self.c=np.mean(y_train)-self.m*np.mean(x_train)
    def predict(self,x_test):
        return self.m*x_test+self.c

```

### Multiple Linear Regrassion
##### Work with more than one input columns(pridictors) or output columns.


```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

```

```python
x,y=make_regression(n_samples=100,n_features=2,n_informative=2,n_targets=1,noise=50)
```

```python
df=pd.DataFrame({'feature1':x[:,0],
                 'feature2':x[:,1],
                 'target':y
                 })
```

```python
df
```

```python
fig=px.scatter_3d(df,x="feature1",y="feature2",z="target")
fig.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pre=lr.predict(x_test)
```

```python
print("mae",mean_absolute_error(y_test,y_pre))
print("mse",mean_squared_error(y_test,y_pre))
print("R2 score",r2_score(y_test,y_pre))
```

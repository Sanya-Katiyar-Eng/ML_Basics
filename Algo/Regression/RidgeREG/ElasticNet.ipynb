### ElasticNet Regression 
##### Combination of both Ridge or Lasso Regression
use when data is too large or data have more feature

Ridge reg : L=Σ(yᵢ − ȳ)²+λ|w²|

Lasso reg : L=Σ(yᵢ − ȳ)²+λ|w|

ElasticNet reg : L=Σ(yᵢ − ȳ)²+a|w²|+b|w|

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.metrics import r2_score
```

```python
#Load Dataset
x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

### Linear regression

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
y_pre=lr.predict(x_test)
r2_score(y_test,y_pre)
```

### Ridge regression

```python
rr=Ridge(alpha=0.1)
rr.fit(x_train,y_train)
y_pre=rr.predict(x_test)
r2_score(y_test,y_pre)
```

### Lasso Regression

```python
llr=Lasso(alpha=0.1)
llr.fit(x_train,y_train)
y_pre=llr.predict(x_test)
r2_score(y_test,y_pre)
```

### ElasticNet Regression

```python
er=ElasticNet(l1_ratio=0.9,alpha=0.005)
er.fit(x_train,y_train)
y_pre=er.predict(x_test)
r2_score(y_test,y_pre)
```

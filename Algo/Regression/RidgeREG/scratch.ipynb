## Ridge Regression from Scratch
1. for 2D
for scratch farmula β₁ = Σ(xᵢ − x̄)(yᵢ − ȳ) / [Σ(xᵢ − x̄)² + λ]
where x̄ = mean of x
      ȳ= mean of y
      λ = Regularization Parameter



      
2. for multiD
for that β̂ = (XᵀX + λI)⁻¹Xᵀy
where X = Feature Matrix
      y = Target Vector
      β = Coefficient Vector
      λ = Regularization Parameter
      I = Identity Matrix


Identity Matrix as Example.................................
I =
[1 0 0
 0 1 0
 0 0 1]

1. For 2D   first using Sklearn

```python
from sklearn.datasets import make_regression
x,y=make_regression(n_features=1,n_targets=1,n_informative=1,noise=20,n_samples=100,random_state=13)
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
print(x_train.shape)
print(x_test.shape)
```

```python
# Performing Linear first for comparision
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
print("coef_ : ",lr.coef_)
print('intercept : ',lr.intercept_)
```

```python
#performing Ridge on alpa 10
from sklearn.linear_model import Ridge
rr1=Ridge(alpha=10)
rr1.fit(x_train,y_train)
print("coef_ : ",rr1.coef_)
print("intercet_ : ",rr1.intercept_)
```

```python
#performing Ridge on alpa 100
from sklearn.linear_model import Ridge
rr2=Ridge(alpha=100)
rr2.fit(x_train,y_train)
print("coef_ : ",rr2.coef_)
print("intercet_ : ",rr2.intercept_)
```

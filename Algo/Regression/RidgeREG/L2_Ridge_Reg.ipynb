```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
x,y=make_regression(n_samples=100,n_features=1,n_targets=1,n_informative=1,noise=23,random_state=42)
plt.scatter(x,y)
```

```python
# performing Linear regression first 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
lr=LinearRegression()
lr.fit(x_train,y_train)
y_pre=lr.predict(x_test)
print("coef_ : ",lr.coef_)
print("intercept : ",lr.intercept_)
print("r2_score : ",r2_score(y_pre,y_test))
print("Mean absolute error : ",mean_absolute_error(y_pre,y_test))

```

```python
# Performing lasso regression
from sklearn.linear_model import Lasso
alpha=[1,10,100]
for i in range(len(alpha)):
    l=Lasso(alpha=alpha[i])
    l.fit(x_train,y_train)
    y_pre=l.predict(x_test)
    print("coef_ : ",l.coef_)
    print("intercept : ",l.intercept_)
    print("r2_score : ",r2_score(y_pre,y_test))
    print("Mean absolute error : ",mean_absolute_error(y_pre,y_test))
```

Linear Regression:

“Prediction ke liye best coefficients find karo.”

“Prediction ke saath-saath unnecessary coefficients ko zero karke model ko simpler banao.”
